```python
def words_count(filepath, words_list):
    with open(filepath) as file:
        text = file.read()

    n = 0
    for word in text.split():
        # Count the number of times the words in the list appear
        if word.lower() in words_list:
            n += 1
    return n
```

```python
some_list = ["This","is","a","bad",",","very","bad",",","list","of","random","words"]

# Create a list of words - function will be used to count words of a filename
with open('wordfile.txt',"w") as f:
    f.write(" ".join(some_list))

# Test it
words_count('wordfile.txt', ['bad','words'])
```
