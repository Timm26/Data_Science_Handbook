# Introduction to Object-Oriented Programming

## 1. What is OOP
1. Procedual Programming
- Code as sequence of steps
- Great for data analysis

2. Object-oriented programming
- Code as interactions of objects
- Great for building software/ Maintainable and reusable code
- _Class_ = a blueprint for objects outlining possible states and behaviors 
- _Object_ = data + functionality. An instance of a class
- _State_ = Data associated with an object, assigned through **attributes**
- _Behavior_ = an object's functionality, defined through **methods**

#### Classes in Python
- Python objects of the same type behave in the same way
- `lists` are a class
1. Created with comma-separated values `[1,2,3,4,5]`
2. Share the same methods, e.g. `.append()`
#### Attributes and methods
- State <-> attributes (`.shape`)
- Behavior <-> methods (`.head()`)
- Displaying all attributes and methods `dir(`[1,2,3,4,5]`)` or `dir(`list`)`

### 1.1 Class anatomy: attributes and methods
- `class <name>` starts a definition
- Use `self` as the first argument in method defintion

```python
class Customer:
    def identity(self, name):
        print("I am Customer " + name)

c_one = Customer() # Create an object of class Customer
c_two = Customer()

c_one.identity("Timm") # Ignore `self` when calling a method on an object
```

Add an attribute to class:

```python
class Customer:
    # Set the name attribute of an object to new_name
    def set_name(self, new_name):
        # Create an attr by assigning a value
        self.name = new_name # will create .name when set_name is called
        
    # Using .name from the object it*self*
    def identify(self):
        print("I am customer " + self.name)
        
# Create an object        
cust = Customer() # .name doesn't exist here yet
# .name is created and set to "Timm Rahrt"
cust.set_name("Timm Rahrt")
print(cust.name)
cust.identify()
```

#### Create a counter class

```python
class Employee:
    def set_name(self, new_name):
        self.name = new_name
        
    def set_salary(self, new_salary):
        self.salary = new_salary
        
    def give_raise(self, amount):
        self.salary = self.salary + amount
        
emp = Employee()
emp.set_name("Timm")
emp.set_salary(100000)
print(f"Old sal {emp.salary}")
emp.give_raise(50000)
print(f"{emp.name}; New sal: {emp.salary}")
```

### 1.2 Class anatomy: the __init__ constructor
- Add data to object when creating it
- **Constructor**: `__init__()` method is called every time an object is created
- Create attributes in the constructor! Attributes are created when object is created

```python
class Calculator:
  def __init__(self, num_one, num_two):
    self.num_one = num_one
    self.num_two = num_two
  
  # Create the addition method
  def addition(self):
    return self.num_one + self.num_two
  
  # Create the subtraction method
  def subtraction(self):
    return self.num_one - self.num_two
  
  # Create the multiplication method
  def multiplication(self):
    return self.num_one * self.num_two

calc = Calculator(5,10)
print(calc.addition(), calc.subtraction(), calc.multiplication())
print(calc.num_one, calc.num_two)
```

### 1.3 Best practices

3. Keep `self` as `self`
4. Use docstrings

## 2. Inheritance and Polymorphism

1. **Encapsulation**: Bundling of data and methods
2. **Inheritance**: Extending the functionality of existing code
3. **Polymorphism**: Creating a unified interface

```python
# Example of Encapsulation
class Employee:
    MIN_SALARY = 30000
    def __init__(self, name, salary):
        self.name = name
        # Use class name to access class attr
        if salary >= Employee.MIN_SALARY:
            self.salary = salary
        else:
            self.salary = Employee.MIN_SALARY
emp1 = Employee("John", 40000)
emp2 = Employee("Jane", 60000)
print(emp1.MIN_SALARY, emp2.MIN_SALARY)
```

```python
# Modifying class-level attributes
emp1.MIN_SALARY = 50000
print(emp1.MIN_SALARY, emp2.MIN_SALARY)
```

#### Why use class attributes? Global constants related to the class

- WHen to use class methods: Alternative constructors, Methods that don't require instance-level attr, Restricting to a single instance(object) of a class

```python
# Example 1
class Person:
    CURRENT_YEAR = 2025
    def __init__(self,name,age):
        self.name = name
        self.age = age
    # Add a class method decorator
    @classmethod
    # Define the from_birth_year method
    def from_birth_year(cls, name, birth_year):
        age = Person.CURRENT_YEAR - birth_year # Create age
        return cls(name,age) # Return the name and age

bob = Person.from_birth_year("Bob",1990)
bob.age
```

```python
# Example 2
class BetterDate:
  def __init__(self, year, month, day):
    self.year, self.month, self.day = year, month, day
    
  # Define a class method from_str
  @classmethod
  def from_str(cls, datestr):
    # Split the string at "-"
    parts = datestr.split("-")
    year, month, day = int(parts[0]), int(parts[1]), int(parts[2])
    # Return the class instance
    return cls(year, month, day)

# Create the xmas object      
xmas = BetterDate.from_str("2024-12-25")   
print(xmas.year)
print(xmas.month)
print(xmas.day)
```

### Class Inheritance
New class functionality = Old class functionality + extra

```python
# BankAccount: Parent class whose functionality is being extended/inherited
class BankAccount:
    def __init__(self, balance):
        self.balance = balance
    
    def withdraw(self, amount):
        self.balance -= amount
    
# Class inheriting from BankAccount
class SavingsAccount(BankAccount):
    pass
```

```python
# Child class has all of the parent data
savings_acct = SavingsAccount(1000)
print(type(savings_acct))
# Attributes inherited from BankAccount
print(savings_acct.balance)
# Method inherited from BankAccount
savings_acct.withdraw(300)
print(savings_acct.balance)
```

#### Customizing constructors

```python
# Add a first child class
class SavingsAccount(BankAccount):
    # Constructor for SavingsAccount with an additional argument
    def __init__(self, balance, interest_rate):
        # Call the parent constructor using ClassName.__init__()
        BankAccount.__init__(self, balance) # self is a SavingsAccount but also a BankAccount
        # Add more functionality
        self.interest_rate = interest_rate
    
    def compute_interest(self, n_periods = 1): # New functionality
        return self.balance * ((1 + self.interest_rate) ** n_periods - 1)
    
# Using Savings Account to calc the earned interest excl. starting Amount A = P([1+r^n] - 1)
sav_acc = SavingsAccount(balance = 500, interest_rate=0.05)
sav_acc.compute_interest(5)
```

```python
# Add a second child class
class CheckingAccount(BankAccount):
    def __init__(self, balance, limit):
        BankAccount.__init__(self, balance) # Call the ParentClass constructor
        self.limit = limit
    def deposit(self, amount):
        self.balance += amount
    def withdraw(self, amount, fee=0): # New fee argument
        if amount <= self.limit:
            BankAccount.withdraw(self, amount+fee)
        else:
            print("Limit too low!") # Won't run if the condition isn't met

# Using Checking Account
check_acc = CheckingAccount(balance = 400, limit = 30)
check_acc.deposit(50)
check_acc.balance
```

```python
check_acc.withdraw(50) # This will be ignored due to set limit
check_acc.withdraw(20, fee = 5) # This will be executed
check_acc.balance
```

#### 3.1.1 The `__eq()__` method
- `__eq__()` is called when 2 objects of a class are compared using `==`
- Accepts 2 arguments, `self` and `other` - objects to compare
- Returns a Boolean

```python
# amend the class
class Customer():
    def __init__(self, name, balance):
        self.name, self.balance = name, balance
    
    def check_balance(self):
        print(f"Customer {self.name} has {self.balance}$ in his account.")
    # NEW: Will be called when == is used
    def __eq__(self, other):
        print("__eq__() is called") # Printout
        # Returns True if all attr match
        return (self.name == other.name) and (self.balance == other.balance)\
            and (type(self) == type(other)) # make sure the class is the same

# Check again - real example should have an acc_id var as name and balance can techn. match for 2 diff. cust.
cust_one = Customer("Timm Rahrt", 5000)
cust_one.check_balance()
cust_two = Customer("Timm Rahrt", 5000)
cust_two.check_balance()
cust_one == cust_two
```

### 3.2 Inheritance comparison and string representation
- `__str__()`: informal, for end user - returns an object in string format
- `__repr__()`: formal, for developer - shows the object in a format representing how it was made

#### 3.2.1 `__repr__()`

```python
class Customer:
    def __init__(self, name, balance):
        self.name = name
        self.balance = balance
    
    def __repr__(self):
        # Notice the '...' around name
        return f"Customer('{self.name}', {self.balance})"
    
cust = Customer("Maryam Azar", 3000)
# Will implicitly call __repr__()
cust
```

#### 3.2.2 `__self__()`

```python
class Customer:
    def __init__(self, name, balance):
        self.name = name
        self.balance = balance
    
    def __str__(self):
        cust_str = f"""
Customer:
    name: {self.name}
 balance: {self.balance}
        """
        return cust_str

cust = Customer("Maryam Azar", 3000)
# Will implicitly call __str__()
print(cust)
```

## 4. Exeptions
### 4.1 Custom exceptions
- Inherit from `Exception` or one of its subclasses
- Usually an empty class

```python
class BalanceError(Exception):
    pass

class Customer():
    def __init__(self, name, balance):
        if balance < 0:
            raise BalanceError("Balance has to be non-negative!")
        else:
            self.name = name
            self.balance = balance
```

```python
# Exception in constructor
try:
    cust = Customer("Larry Torres", -100)
except BalanceError:
    cust = Customer("Larry Torres", 0)

cust.balance
```
