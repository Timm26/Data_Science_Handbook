## 1. Functions in Python

### 1.1 Returning multiple values

```python
def raise_both(value1, value2): # Function header
    """Raise value1 to the power of value2 and vice versa. # Docstring
    """
    new_value1 = value1 ** value2   # Function Body
    new_value2 = value2 ** value1
    return (new_value1,new_value2) # Return value(s)

# Use it
raise_both(3,4)
    
```

### 1.2 Nested function

```python
def raise_val(n):
    "Return inner function"
    def inner(x):
        "Raise x to the power of n"
        raised = x ** n
        return raised
    return inner

square = raise_val(2)
cube = raise_val(3)
print(square(2), cube(4))
```

### 1.3 Flexible arguments
#### 1.3.1 *args

```python
def add_all(*args):
    "Sum all values in *args together"
    sum_all = 0
    for num in args:
        sum_all += num
    return sum_all

# Use it
print(add_all(2,3))
print(add_all(5,10,15,20,25))
```

#### 1.3.2 **kwargs

```python
def print_all(**kwargs):
    "Print out key-value pairs in **kwargs"
    # Print out the key-value pairs
    for key, value in kwargs.items():
        print(key + ": " + value)
        
# Use it
print_all(Name = "Timm", Occupation = "Student", Age = "26")
```

## 2. Scope
- Part of the program where an object or name may be accessible
- *Global scope* - defined in the main body of the script
- *Local scope* - defined inside a function
- *Built-in scope* - names in the pre-defined built-ins module

```python
# Global vs. local scope
new_val = 10

def square(value):
    "Returns the square of a number."
    global new_val
    new_val = new_val ** 2
    return new_val

print(f"Before calling the function, new_val: {new_val}")
# Check function
print(f"Function: {square(3)}")
# Check new_val
print(f"After calling the function, new_val: {new_val}")
```

### 3.1 Anonymous functions

```python
# Take the square root of all numbers in the list
nums = [21,52,35,73,14]

sqrt_all = map(lambda num: num ** 0.5, nums)
print(list(sqrt_all))
```

## 4. Error Handling

### 4.1 Errors and exceptions
- Exceptions - caught during execution
- Catch exceptions with try-except clause

```python
def sqrt(x):
    "Returns square root for a number."
    if x < 0:
        raise ValueError("x must be non-negative")
    try:
        return x ** 0.5
    except TypeError:
        print('x must be an int or float')
        
# Use it
sqrt(81)
```

## 5. Iterators

### 5.1 Iterating over different objects
- **_Iterable_:** Lists, Strings, Dictionaries, File connections
- An object with an associated *iter()* method
- Applying *iter()* to an iterable creates an iterator
- **_Iterator_:**: Produces next value with *next()*

```python
# Iterate over a list
employees = ['Steve', 'Marc', 'Otto']
for employee in employees:
    print(employee)

# Iterate over a string
for letter in "Course":
    print(letter)
    
# Iterate over a range
for i in range(3):
    print(i)
```

### 5.2 Iterating over iterables: _next()_
- *iter()*

```python
word = 'Yo'
it = iter(word)
next(it)
```

```python
next(it)
```

```python
word = 'Course'
it2 = iter(word)
print(*it2)
```

### 5.3 Iterating over dictionaries
- Use *.items()* to get key, value pairs

```python
students = {"Timm":"Rahrt", "James":"Castro"}
for key, value in students.items():
    print(key, "=", value)
```

### 5.4 Using *enumerate()*
- Returns index, value pairs

```python
avengers = ["hawkeye", "iron man", "thor", "quicksilver"]
e = enumerate(avengers)
list(e) # Display results
```

```python
# Unpack the results
for index, value in enumerate(avengers):
    print(index, value )
```

### 5.5 Zip two lists together using *zip()*

```python
# Using zip
first_names = ["Timm", "Emma", "Sarah", "Kurt"]
last_names = ["Rahrt", "Watson", "Kreutzer", "Kurtiger"]
list(zip(first_names, last_names))
```

## 6. List & Dict comprehension
- Collapse _for loops_ for building lists into a single line

### 6.1 List comprehension with *range()*

```python
[num for num in range(11)]
```

#### 6.1.1 Nested List comprehension
- Tradeoff: Readability

```python
# Returns a list of tuples
[(num1, num2) for num1 in range(0,2) for num2 in range(8,11)]
```

#### 6.1.2 Conditionals in comprehensions

```python
# Finding uneven numbers
[num ** 2 for num in range(10) if num % 2 != 0]
```

```python
# Conditionals on the output expression
[num ** 2 if num % 2 == 0 else 99 for num in range(10)] # Even number or 99
```

### 6.2 Dict comprehensions
- Create dictionaries
- Use {} instead of []

```python
# Dictionary of pos:neg nubers
pos_neg = {num: -num for num in range(5)}
print(pos_neg)
```

## 7. Generator expression

### 7.1 Generator expression
- Returns a generator object
- Can be iterated OverflowError
- Doesn't save the result in memory -> Lazy evaluation

```python
result = (num for num in range(5))
for num in result:
    print(num)
```

```python
result2 = (num for num in range(6))
print(next(result2))
print(next(result2))
print(next(result2))
```

### 7.2 Build a generator function
- Produces generator objects when called
- Yields a sequence of values instead of returning a single value
- Generates a yield keyword

```python
def num_sequence(n):
    "Generate values from 0 to n"
    i = 0
    while i < n:
        yield i
        i += 1
```

```python
result3 = num_sequence(5)
print(next(result3))
print(next(result3))
print(next(result3))
```
