# Joining Data with pandas

```python
import pandas as pd
import numpy as np
import pickle
census = pd.read_csv("../ZDatasets/Chicago_census.csv")
wards = pd.read_csv("../ZDatasets/Chicago_wards.csv")
# Load the pickle file licenses
with open("../ZDatasets/licenses.p", "rb") as f:
    licenses = pickle.load(f) # could also use pd.read_pickle
# Correct the ward column to an integer
licenses['ward'] = licenses['ward'].astype('int64')
movies_to_genres = pd.read_pickle('../ZDatasets/movie_to_genres.p')
```

## 1. Data Merging Basics

### 1.1 Inner Join
Keeps only the rows where the value of 'ward' exists in both DataFrames.

```python
wards_census = wards.merge(census, on = 'ward', suffixes=("_ward","_cen"))
wards_census.head()
```

### 1.2 One to many relationships
Every row in the left table is related to only one row in the right table.

### 1.3 One to many relationships
Every row in the left table is related to one or more rows in the right table.

```python
# Example
ward_licenses = wards.merge(licenses, on = 'ward', suffixes = ('_ward','_lic'))
print(wards.shape)
print(ward_licenses.shape)
```

```python
### 1.4 Merging multiple tables
wards.merge(licenses, on = 'ward', suffixes=("_wards","_lic")).merge(census, on='ward', suffixes = ("_wards","_census")).head()
```

## 2. Merging Tables with Different Join Types

### 2.1 Left join
Keeps all the rows from the left DataFrame, and only the matching rows from the right DataFrame.

```python
# Get the data
movies = pd.read_pickle("../ZDatasets/movies.p")
taglines = pd.read_pickle("../ZDatasets/taglines.p")
```

```python
# Merge with left join
movies_taglines = movies.merge(taglines, on = 'id', how = 'left')
movies_taglines.head()
```

### 2.2 Right join
Keeps all the rows from the right DataFrame, and only the matching rows from the left DataFrame.

```python
movies_taglines_rj = movies.merge(taglines, on = 'id', how = 'right')
movies_taglines_rj.head()
```

### 2.3 Outer join
Keeps all rows from all tables. Will use 'NaN' for missing values.

```python
movies_taglines_oj = movies.merge(taglines, on = 'id', how = 'outer')
movies_taglines_oj.head()
```

### 2.4 Self join
Merging a table to itself.

```python
# Get the data
sequels = pd.read_pickle("../ZDatasets/sequels.p")
```

```python
original_sequels = sequels.merge(sequels, left_on = 'sequel', right_on = 'id', suffixes = ('_org','_seq'))
original_sequels.head()
```

```python
# Show sequels
original_sequels.loc[:,['title_org','title_seq']].head()
```

### 2.5 Index merge

```python
movies_genres = movies.merge(movies_to_genres, left_on = "id", right_on = "movie_id")
movies_genres.head()
```

## 3. Advanced Merging and Concatenating

### 3.1 Filtering joins
- Mutating joins: Combines data from two tables based on matching observations in both tables.
- Filtering joins: Filter observations from table based on whether or not they match an observation in another table.

### 3.2 Semi joins
Returns the intersection, only columns from the left table and not the right. Doesn't return duplicates.

```python
movies_genres = movies.merge(movies_to_genres, left_on ="id", right_on = "movie_id")
mov = movies[movies['id'].isin(movies_genres['id'])]
mov.head()
```

### 3.2 Anti join
Returns only columns from the left table and not the right, it excludes the intersection.

```python
movies_genres2 = movies.merge(movies_to_genres, left_on = 'id', right_on = "movie_id", how = 'left', indicator= True) # indicator for '_merge' column
mov_grid = movies_genres2.loc[movies_genres2['_merge'] == 'left_only', 'id'] # set up the left only condition
mov2 = movies[movies['id'].isin(mov_grid)] # filter for left_only rows
mov2.head()
```

### 3.3 Concatenate two tables (vertically or horizontally)
_pd.concact()_: Concatenate tables verical and horizontal (_axis=0_, vertical)

```python
pd.concat([movies,movies_to_genres],axis=0,
          keys = ['movies','movies_to_genres']).head()
```

## 4. Merging Ordered and Querying

### 4.1 Using *merge_ordered()*
- Column(s) to join on _on_, *left_on* and *right_on*
- Type of join *how* (left, right, inner, outer)
- _suffixes_ for overlapping column names
- Calling the function: *pd.merge_ordered(df1, df2)*

### 4.2 Using *merge_asof()*
- Match on the nearest key column and not exact matches
- Use when data samples from a process and when developing a trainin set (no data leakage)

### 4.3 *query()*
- Accepts an input string
- Input string is similar to statement after WHERE clause in SQL statement

```python
movies.query("release_date < '1999-07-03'").head()
```
