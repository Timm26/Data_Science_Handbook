```python
import pandas as pd
import numpy as np
import pickle
census = pd.read_csv("../ZDatasets/Chicago_census.csv")
wards = pd.read_csv("../ZDatasets/Chicago_wards.csv")
# Load the pickle file licenses
with open("../ZDatasets/licenses.p", "rb") as f:
    licenses = pickle.load(f) # could also use pd.read_pickle
# Correct the ward column to an integer
licenses['ward'] = licenses['ward'].astype('int64')
movies_to_genres = pd.read_pickle('../ZDatasets/movie_to_genres.p')
```

## 1. Data Merging Basics

### 1.1 Inner Join
Keeps only the rows where the value of 'ward' exists in both DataFrames.

```python
wards_census = wards.merge(census, on = 'ward', suffixes=("_ward","_cen"))
wards_census.head()
```

### 1.2 One to many relationships
Every row in the left table is related to only one row in the right table.

### 1.3 One to many relationships
Every row in the left table is related to one or more rows in the right table.

```python
# Example
ward_licenses = wards.merge(licenses, on = 'ward', suffixes = ('_ward','_lic'))
print(wards.shape)
print(ward_licenses.shape)
```

```python
### 1.4 Merging multiple tables
wards.merge(licenses, on = 'ward', suffixes=("_wards","_lic")).merge(census, on='ward', suffixes = ("_wards","_census")).head()
```

## 2. Merging Tables with Different Join Types

### 2.1 Left join
Keeps all the rows from the left DataFrame, and only the matching rows from the right DataFrame.

```python
# Get the data
movies = pd.read_pickle("../ZDatasets/movies.p")
taglines = pd.read_pickle("../ZDatasets/taglines.p")
```

```python
# Merge with left join
movies_taglines = movies.merge(taglines, on = 'id', how = 'left')
movies_taglines.head()
```

### 2.2 Right join
Keeps all the rows from the right DataFrame, and only the matching rows from the left DataFrame.

```python
movies_taglines_rj = movies.merge(taglines, on = 'id', how = 'right')
movies_taglines_rj.head()
```

### 2.3 Outer join
Keeps all rows from all tables. Will use 'NaN' for missing values.

```python
movies_taglines_oj = movies.merge(taglines, on = 'id', how = 'outer')
movies_taglines_oj.head()
```

### 2.4 Self join
Merging a table to itself.

```python
# Get the data
sequels = pd.read_pickle("../ZDatasets/sequels.p")
```

```python
original_sequels = sequels.merge(sequels, left_on = 'sequel', right_on = 'id', suffixes = ('_org','_seq'))
original_sequels.head()
```

```python
# Show sequels
original_sequels.loc[:,['title_org','title_seq']].head()
```

### 2.5 Index merge

```python
movies_genres = movies.merge(movies_to_genres, left_on = "id", right_on = "movie_id")
movies_genres.head()
```
