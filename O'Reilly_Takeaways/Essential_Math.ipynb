### 1.10 Derivatives
$f(x) = x^2$

$\frac{d}{dx}f(x) = \frac{d}{dx}x^2 = 2x$
- Tells the slope of a function and is useful to measure the rate of change at any point in a function
- when the slope is 0, we are at a minimum or maximum of an output variable
- calculate the slope

$m = \frac{y_2 - y_1}{x_2 - x_1}$

```python
# Calculate the derivative
def derivative_x(f, x, step_size):
    m = (f(x + step_size) - f(x)) / ((x + step_size) - x)
    return m

# My function
def my_function(x):
    return x**2

# Use
slope_at_2 = derivative_x(my_function, 2, .00001)
print(slope_at_2)
```

## 1.12 Chain rule
For a given function y (with input varibale x) composed into another function z (with input variable y), we can find the derivative of z with respect to x by multiplying the two respective derivatives together:

$\frac{dz}{dx} = \frac{dz}{dy} * \frac{dy}{dx}$

Example -> derivative of $y = x^2 + 1$ and $z = y^3 - 2$:

```python
from sympy import *
x,y = symbols('x y')
# derivative of first function
_y = x**2 +1
dy_dx = diff(_y)
# derivative for second function
z = y **3 -2
dz_dy = diff(z)

(dy_dx * dz_dy).subs(y, _y)
```

# 2. Proability
_Probability_: Quantifying predictions of events yet to happen. Purely theoretical of how likely an event is to happen.
### 2.1 Probability Basics
How strongly we believe an event will happen, often expressed as a percentage.

- $P(X)$ where $X$ is the event of interest
- $P(X)$ = 0.70

_Likelihood_: Measuring the frequency of events that already occurred.

$P(\text{not X}) = 1 - 0.70 = 0.30$

_Odds_: $O(X) = \frac{P(X)}{1 - P(X)}$

### 2.2 Single probability
_Marginal probability_ = $P(X)$ = single probability of an event

### 2.3 Joint Probability
$P(\text{A AND B}) = P(A) * P(B)$

### 2.4 Union Probability
Probability of getting event A or B? Mutually exclusive events, events that cannot occur simultaneously. Rolling a die once.

$P(\text{4 OR 6}) = P(A) + P(B) = \frac{1}{6} + \frac{1}{6} = \frac{1}{3}$ 

Nonmutally exclusive events - can occur simultaneously:

$P(\text{A OR B}) = P(A) + P(B) - P(A) * P(B)$

### 2.5 Conditional Probability
$P(\text{A GIVEN B}) = P(A|B)$

### 2.6 Bayes' Theorem
$P(A|B) = \frac{P(B|A)*P(A)}{P(B)}$

```python
# Example coffee and cancer
p_coffee_drinker = 0.65
p_cancer = 0.005
p_coffee_drinker_given_cancer = 0.85
# Calculation
p_cancer_given_coffee_drinker = p_coffee_drinker_given_cancer * p_cancer / p_coffee_drinker
print(p_cancer_given_coffee_drinker)
```

We multiply a conditional probability by the probability of the condition:

$P(\text{A and B}) = P(A|B) * P(B)$ 

_or_

$P(\text{A and B}) = P(B|A) * P(A)$

## 3.1 Descriptive Statistics

### 3.1.1 Mean and Weighted Mean
Sample Mean: $\sum{\frac{x_i}{n}}$

Population Mean: $\sum{\frac{x_i}{N}}$

_Mean_: Average of a set of values:

```python
# Mean
def mean(xs):
    return sum(xs) / len(xs)

list1 = [1,3,5,8]
mean(list1)
```

_Weighted Mean_: $\frac{\sum{(x_i * w_i)}}{w_1 + ... + w_n}$

```python
# Weighted Mean
sample = [90, 80, 63, 87]
weights = [.20, .20, .20, .40]
weighted_mean = sum(s * w for s,w in zip(sample, weights)) / sum(weights)
print(weighted_mean)
```

### 3.1.2 Median
Middlemost value in a set of ordered values.

```python
def median(xs):
    ordered = sorted(xs)
    n = len(ordered)
    mid = int(n / 2) - 1 if n%2 == 0 else int(n/2)
    if n%2 == 0:
        return (ordered[mid] + ordered[mid+1]) / 2.0
    else:
        return ordered[mid]
    
print(median(list1))
```

### 3.1.3 Mode
Most frequently occurring set of values.

```python
# Calcuating the mode
from collections import defaultdict
sample = [1,3,2,5,7,0,2,3]

def mode(values):
    counts = defaultdict(lambda: 0)
    
    for s in values:
        counts[s] += 1
    
    max_count = max(counts.values())
    modes = [v for v in set(values) if counts[v] == max_count]
    return modes

print(mode(sample))
```

### 3.1.4 Population Variance and Standard Deviation
_Population Variance_: 
    
$\sigma^2 = \frac{(x_1 - mean)^2 + (x_2 - mean)^2 + ... + (x_n - mean)^2}{N}$

```python
# Variance
data = [0,1,5,7,9,10,14]

def variance(values):
    mean = sum(values) / len(values)
    _variance = sum((v - mean)**2 for v in values) / len(values)
    return _variance

print(variance(data))
```

_Standard Deviation:_

$\sigma = \sqrt{\frac{\sum{(x_1 - \mu)^2}}{N}}$

```python
# Standard Deviation
def std_dev(values):
    mean = sum(values) / len(values)
    _variance = sum((v - mean) **2 for v in values) / len(values)
    return np.sqrt(_variance)

print(std_dev(data))
```

### 3.1.5 Sample Variance and Standard Deviation
By counting values short of one item in the divisor, we increase the variance and therefore capture greater uncertainty in our sample.

$s^2 = \frac{\sum{(x_i - \hat{x}})^2}{n - 1}$

$s = \sqrt{\frac{\sum{(x_i - \hat{x})^2}}{n - 1}}$

```python
# Calculating standard deviation for a sample
def variance(values, is_sample: bool = False):
    mean = sum(values) / len(values)
    _variance = sum((v - mean)**2 for v in values) / (len(values) - (1 if is_sample == True else 0))
    return _variance

def std_dev(values, is_sample:bool = False):
    return sqrt(variance(values, is_sample))

print(f"Variance:, {variance(data, is_sample=True)}")
print(f"Standard Deviation: {std_dev(data, is_sample=True)}")
```

### 3.1.6 The Normal Distribution

- The "tails" are the least likely outcomes and approach zero but never touch zero

```python
from scipy.stats import norm
mu = 0
sigma = 1
x = np.linspace(-4,4,500)
y = norm.pdf(x, mu, sigma) # definition below
# Plot
plt.plot(x,y,lw=2)
plt.show()
```

#### The Probability Function (PDF)
Used to look up the likelihood at that given value.

$f(x) = \frac{1}{\sigma}* \sqrt{2 \pi} * e^{-\frac{1}{2} (\frac{x - \mu^2}{\sigma})}$

#### The Cumulative Distribution Function (CDF)
The y-axis is the likelihood, use CDF to calculate the probability of that given range. Let's find the probability a golden retriever weights between 62 and 66 pounds:

```python
mean = 64.43
std_dev = 2.99
x = norm.cdf(66, mean, std_dev) - norm.cdf(62, mean, std_dev)
print(f"{np.round(x,3)*100}%")
```

#### Z-Scores
Common to rescale a normal distribution so that the mean is 0 and the standard deviation is 1, which is known as the _standard normal distribution_.
- Makes it easier to compare the spread of one normal distribution to another normal distribution, even if they have different means and variances.
- Z-score expresses all x-values into a basic scaling: $z = \frac{x - \mu}{\sigma}$
- Getting the x-value of a z-score: $x = z*\sigma + \mu$

```python
# Example
mu_A = 140000 # Mean home neigborhood A 
mu_B = 800000 # Mean home neigborhood B
sigma_A = 3000 # Standard Deviation neighborhood A
sigma_B = 10000 # Standard Deviation neighborhood B
x_A = 150000 # Home A
x_B = 815000 # Home B
```

```python
# Z-value
z_A = (x_A - mu_A) / sigma_A
z_B = (x_B - mu_B) / sigma_B
print(f"Z-score of house in Neigborhood A: {z_A}")
print(f"Z-score of house in Neigborhood B: {z_B}")
```

House in neighborhood A is much more expensive with a Z-score of 3.33.

```python
# Using python
def z_score(x, mean, std):
    return (x-mean)/std

def z_to_x(z, mean, std):
    return (z*std) + mean

# Z-score
print(z_score(x_A, mu_A, sigma_A))
# X-value
print(z_to_x(z_A, mu_A, sigma_A))
```

## 3.2 Inferential Statistics

### 3.2.1 Central Limit Theorem
1. The mean of the sample means is equal to the population mean
2. If the population is normal, then the sample means will be normal
3. If the population is not normal, then the sample means will be normal
4. The standard deviation of the sample means equals the populations standard deviation divided by the square root of $n$:
    
sample standard deviation = $\frac{\text{population standard deviation}}{\sqrt{\text{sample size}}}$

Useful to infer insights about populations based on samples, even for nonnormal populations.
- n > 30 # thirty-one is the textbook number in statistics
- When you have fewer than 31 items you have to rely on the _T-distribution_, which has increasingly fatter tails the smaller your sample size to increase uncertainty

```python
# Example
import plotly.express as px

sample_size = 31
sample_count = 1000

# CLT 1000 samples with each n = 31
# random numbers between 0.0 and 1.0 - uniform distribution
x_values = [(sum([random.uniform(0.0, 1.0) for i in range(sample_count)]) / sample_count) for _ in range(sample_count)]
y_values = [1 for _ in range(sample_count)]

px.histogram(x=x_values, y=y_values, nbins=20).show()
```

### 3.2.2 Confidence Intervals
Is a range calculation showing how confidently we believe a sample mean/ proportion/ other parameter falls in a range of the population mean.

1. Define your level of confidence (LOC) - 5% is commonly used
2. Calculate the critical z-value. If $n$ < 31 use T-Distribution.

```python
# Retrieve a critical z-value
from scipy.stats import norm
def critical_z_value(p):
    norm_dist = norm(loc = 0.0, scale = 1.0) # creates a standard normal distribution
    left_tail_area = (1.0 - p) / 2.0 # Compute tail area left 0.025
    upper_area = 1.0 - ((1.0 - p) / 2.0) # Compute tail area right 9.975
    return norm_dist.ppf(left_tail_area), norm_dist.ppf(upper_area) # convert probabilities to z-values

print(critical_z_value(p=.95))
```

3. Compute margin of error $E = \pm z_c \frac{S}{\sqrt{n}}$. Note: The larger n the narrower $E$.
4. 95% confidence interval = $X \pm E$.
5. Interpret results

```python
# Example
from math import sqrt
# def critical_z_value(p): #already defined

def confidence_interval(p, sample_mean, sample_std, n):
    # sample size must be greater than 30
    lower, upper = critical_z_value(p)
    lower_ci = lower * (sample_std/sqrt(n))
    upper_ci = upper * (sample_std/sqrt(n))
    return sample_mean + lower_ci, sample_mean + upper_ci

# Try
print(confidence_interval(p=.95, sample_mean=64.408, sample_std=2.05, n=31))
```

Based on a sample of 31 golden retriever weights with sample mean 64.408 and sample standard deviation of 2.05, I am 95% confident the population mean lies between 63.68 and 65.13.

### 3.2.3 Understanding P-Values
_P-Value_: Probability that something is happening by chance, rather than because of a hypothesized explanation.
- Threshold for statistical significance is a p-value of 5% or less.

_Null Hypothesis_: Variable question had no impact on the experiment and any positive results are just random luck

_Alternative Hypothesis_: Variable in question (Controlled Variable) is causing a positive result.

## 3.3 Hypothesis Testing
- Is a treatment showing a significant result or not?

### 3.3.1 One-Tailed Test
- Typically frame the null and alternative hypothesis using reduce_inequalities
- $H_0$: population mean $\geq$ 18
- $H_1$: population mean $\leq$ 18
- Define significance level - typically 5%

```python
# Cold has 18 day mean recovery, 1.5 std dev
mean = 18
std_dev = 1.5
# What x value as 5% of area behind it?
x = norm.ppf(0.05, mean, std_dev)
print(x)
```

We achieve an average 15.53 or fewer days of recovery time in our sample group, our drug is considered statistically significant enough to have shown an impact.

```python
# Print the one-tailed p-value
# Probability of 16 or less days
p_value = norm.cdf(16, mean, std_dev)
print(p_value)
```

### 3.3.2 Two-Tailed Test
