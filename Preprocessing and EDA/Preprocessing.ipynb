# Data Preprocessing

## 1. Transformation

- Certain statistical methods rely on variables having a linear relationship
- E.g. Correlation coefficient
- Linear Regression

```python
def scatter_plot(data, x, y, ci=None):
    '''Visualise the relationship of two continuous variables and add a trendline'''
    sns.lmplot(x=x, y=y, data=data, ci=ci)
    plt.gcf().set_size_inches(5,3)
    plt.show()
```

```python
scatter_plot(data= Bicycle, x='CT_VOLUME_PMPEAK', y='CT_VOLUME_24HOUR')
```

The plot is right tailed, let's use "Square root transformation" to fix this.

### 1.1 Square Root Transformation
Best used for right-skewed data.

```python
Bicycle['CT_VOLUME_PMPEAK'] = np.sqrt(Bicycle['CT_VOLUME_PMPEAK'])
scatter_plot(data= Bicycle, x='CT_VOLUME_PMPEAK', y='CT_VOLUME_24HOUR')
```
